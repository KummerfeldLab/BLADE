# Visium Artifact Cleanup

Notebook demonstrating how to run the standard Visium artifact removal workflow using `Artifact_remove`.


```python
import os, sys
from pathlib import Path

repo_root = Path('..').resolve()
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

from artifactsRemoval import Artifact_remove, Artifact_remove_HD

data_dir = '/path/to/visium/sample'  # update to your Visium dataset
print('Using data_dir:', data_dir)
print('Contains matrix.mtx.gz?', os.path.exists(os.path.join(data_dir, 'filtered_feature_bc_matrix', 'matrix.mtx.gz')))
print('Contains tissue_positions.csv?', os.path.exists(os.path.join(data_dir, 'spatial', 'tissue_positions.csv')))
```

```python
visium = Artifact_remove(dir=data_dir)

print('Barcodes:', visium.Tissue.barcode_list.shape[0])
print('Features:', visium.Tissue.feature_list.shape[0])
print('Positions:', visium.Tissue.tissue_position.shape[0])
```

## Outlier Detector Check

Reuse the HD outlier detector on aggregated counts to confirm fast convergence.


```python
import time, warnings

summary_df = visium.get_sum()
start = time.perf_counter()
with warnings.catch_warnings(record=True) as caught:
    warnings.simplefilter('always', RuntimeWarning)
    outliers = Artifact_remove_HD.outlier(summary_df)
duration = time.perf_counter() - start
rate = len(outliers) / len(summary_df) if len(summary_df) else 0.0

print(f'Detected {len(outliers)} outlier spots ({rate:.2%}) in {duration:.2f} seconds')
if caught:
    for w in caught:
        print(f'Warning: {w.message}')
```

```python
# Remove artifacts
visium.remove_border()
visium.remove_edge(distance=2)
visium.remove_malfunction(use_clusters=False)  # standard workflow
visium.review_removing()
```

```python
# Save outputs
output_dir = os.path.join(data_dir, 'blade_outputs')
os.makedirs(output_dir, exist_ok=True)
visium.save(output_dir)
```
