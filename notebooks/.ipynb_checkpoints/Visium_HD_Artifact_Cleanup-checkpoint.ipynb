# Visium HD Artifact Cleanup (BLADE)

This notebook demonstrates how to use the HD artifact removal with KDTree neighbors and HDF5 I/O.

It loads a Visium HD binned output directory, computes neighborhoods, removes border/edge/malfunction spots, and saves filtered outputs (positions CSV and filtered HDF5).

```python
import os, sys
from pathlib import Path

# Path to this repo so we can import the package in-place
repo_dir = '/users/2/wan00232/ondemand/jupyter/Visium/QC/Gitversion/BLADE'
if repo_dir not in sys.path:
    sys.path.insert(0, repo_dir)

data_dir = '/users/2/wan00232/data_folder/Visium_HD/binned_outputs/square_008um'
positions_path = os.path.join(data_dir, 'spatial', 'tissue_positions.parquet')
# Prefer cell_matrix for HD; fallback to bc_matrix for classic
h5_default = os.path.join(data_dir, 'filtered_feature_cell_matrix.h5')
if not os.path.exists(h5_default):
    h5_default = os.path.join(data_dir, 'filtered_feature_bc_matrix.h5')
print('Data dir:', data_dir)
print('Positions path exists:', os.path.exists(positions_path))
print('H5 exists:', os.path.exists(h5_default))
```

```python
# Import the HD artifact remover
from artifactsRemoval.Artifact_remove_HD import Artifact_remove_HD

# Initialize with 8-neighbor (square) KDTree mode
ar = Artifact_remove_HD(
    data_dir=data_dir,
    h5_path=h5_default,
    positions_path=positions_path,
    neighbor_mode='k',   # 'k' = k-nearest, 'radius' for radius-based
    k_neighbors=8,       # 8-neighbor square topology
    compute_depth_on_init=False  # defer depth map for faster init
)

print('Barcodes:', ar.barcode_list.shape)
print('Features:', ar.feature_list.shape)
print('Positions:', ar.tissue_position.shape)
```

```python
# Quick summary of sequencing depth (sum counts per spot)
sum_df = ar.get_sum()
sum_df.head()
```

```python
# OPTIONAL — Compute tissue depth map (distance in layers from the edge/border)
# Compute explicitly when needed (this can take time on large datasets)
# ar.fn_spot_dis_to_edge()
# print('Computed tissue depth up to level:', ar.tissue_depth)
# ar.tissue_depth_df.head()
```

## Remove Border Spots
Two strategies are supported:
- `extreme`: remove spots within a coordinate width from the min/max edges (default).
- `degree`: remove spots that have fewer than a given number of neighbors (default 8).

```python
# Example 1: Extreme border with exact edge only (distance=0)
ar.remove_border(distance=float(ar._grid_radius) if ar._grid_radius else 0.0, strategy='extreme')

# Example 2: Extreme border with a small coordinate width
ar.remove_border(distance=float(ar._grid_radius) if ar._grid_radius else 0.0, strategy='extreme')

# Example 3: Degree-based border (uncomment to use)
# ar.remove_border(strategy='degree', expected_neighbors=8)
```

## Remove Edge Layers
Removes all spots up to a specified layer depth from the border/edge.

```python
# Remove 2 layers from the edge inward (zeros-based seeding by default)
# If distance > max_depth_limit (default 4), it will clamp and warn.
ar.remove_edge(distance=3)
```

## Remove Malfunction (Cluster-Aware Outliers)
Removes only contiguous clusters of outlier spots (default min size = 15).
Set use_clusters=False to use non-spatial outlier removal.


```python
# Cluster-aware removal (default, min_cluster_size=15)
ar.remove_malfunction(min_cluster_size = 30)

# Optional: non-clustered removal (removes all outliers)
# ar.remove_malfunction(use_clusters=False)

print('Removal steps applied:')
ar.review_removing()
```

## Save Outputs
- Updated tissue positions CSV (with updated in_tissue)
- Filtered 10x-style HDF5 matrix with kept spots only

```python
output_dir = os.path.join(data_dir, 'blade_outputs')
os.makedirs(output_dir, exist_ok=True)

# Save positions CSV
ar.save(output_dir)

# Save filtered HDF5
out_h5 = os.path.join(output_dir, 'filtered_feature_cell_matrix.filtered.h5')
ar.save_hdf5(out_h5)

print('Saved to:', output_dir)
```
