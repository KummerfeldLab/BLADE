# One-button Edge Cleanup

Notebook demonstrating how to run the standard Visium artifact cleanup workflow with the new one-button helper `run_iterative_edge_cleanup`, adapted from the Visium artifact cleanup example notebook.


```python
from artifactsRemoval import run_iterative_edge_cleanup
import os
```

Set your input Visium directory (contains `filtered_feature_bc_matrix/` and `spatial/`) and a folder to write intermediate HDF5 outputs.


```python
data_dir = "/path/to/visium/sample"
output_dir = os.path.join("/tmp", "edge_cleanup_runs")
```

Run the one-button helper. It will:
- optionally remove the annotated border;
- t-test edge vs. interior total counts;
- remove one edge layer while `p < threshold` and save an HDF5 snapshot after each removal.


```python
history = run_iterative_edge_cleanup(
    data_dir=data_dir,
    output_dir=output_dir,
    p_threshold=0.05,
    initial_edge_distance=1,
    max_iterations=5,
    remove_border_first=True,
)
```

Inspect the per-iteration p-values and the saved files under `output_dir`.


```python
print("Iterations and p-values:")
for record in history:
    print(record)

print("\nSaved files:")
for fname in sorted(os.listdir(output_dir)):
    if fname.endswith(".h5"):
        print(fname)
```

You can load the final HDF5 in downstream tools or continue with custom filtering using `Artifact_remove`.
